[Создать базовый класс Task в C#, который будет представлять задачи внутри
проекта. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.]


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Task
{
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public bool IsCompleted { get; protected set; }
    public string AssignedTo { get; protected set; }

    public Task(int taskId, string taskName, string priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        IsCompleted = false;
        AssignedTo = "Не назначен";
    }

    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"Задача '{TaskName}' (ID: {TaskId}) отмечена как выполненная.");
    }

    public virtual string GetTaskDetails()
    {
        return $"ID задачи: {TaskId}\n" +
               $"Название задачи: {TaskName}\n" +
               $"Приоритет: {Priority}\n" +
               $"Статус: {(IsCompleted ? "Выполнена" : "В процессе")}\n" +
               $"Назначена: {AssignedTo}";
    }

    public virtual void ReassignTo(string newAssignee)
    {
        AssignedTo = newAssignee;
        Console.WriteLine($"Задача '{TaskName}' (ID: {TaskId}) переназначена на {newAssignee}.");
    }
}

public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }

    public DelegateTask(int taskId, string taskName, string priority, DateTime dueDate)
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
    }

    public override void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"Делегатская задача '{TaskName}' (ID: {TaskId}) отмечена как выполненная {DateTime.Now:yyyy-MM-dd HH:mm}.");
        Console.WriteLine($"   Срок выполнения был: {DueDate:yyyy-MM-dd}");
        Console.WriteLine($"   {(DateTime.Now <= DueDate ? "Выполнено вовремя" : "Выполнено с опозданием")}");
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() +
               $"\nСрок выполнения: {DueDate:yyyy-MM-dd}\n" +
               $"Дней до срока: {(DueDate - DateTime.Now).Days}";
    }
}

public class TeamTask : Task
{
    public string TeamName { get; set; }

    public TeamTask(int taskId, string taskName, string priority, string teamName)
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
    }

    public override void ReassignTo(string newAssignee)
    {
        string oldAssignee = AssignedTo;
        AssignedTo = newAssignee;
        Console.WriteLine($"Командная задача '{TaskName}' (ID: {TaskId}) переназначена с {oldAssignee} на {newAssignee}.");
        Console.WriteLine($"   Команда: {TeamName}");
        Console.WriteLine($"   Приоритет задачи: {Priority}");
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() +
               $"\nКоманда: {TeamName}";
    }
}

public class ResearchTask : Task
{
    public string DataSource { get; set; }
    public string ResearchArea { get; set; }

    public ResearchTask(int taskId, string taskName, string priority, string dataSource, string researchArea)
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
        ResearchArea = researchArea;
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() +
               $"\nОбласть исследования: {ResearchArea}\n" +
               $"Источник(и) данных: {DataSource}";
    }

    public override void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"Исследовательская задача '{TaskName}' (ID: {TaskId}) отмечена как выполненная.");
        Console.WriteLine($"   Область исследования: {ResearchArea}");
        Console.WriteLine($"   Источник данных: {DataSource}");
        Console.WriteLine($"   Результаты исследования готовы к проверке");
    }
}


Console.WriteLine("".PadRight(60, '='));
Console.WriteLine("Система управления задачами - демонстрация");
Console.WriteLine("".PadRight(60, '='));
Console.WriteLine();

Console.WriteLine("Делегатская задача");
Console.WriteLine("".PadRight(40, '-'));

DelegateTask delegateTask = new DelegateTask(
    101, 
    "Создать документацию проекта", 
    "Высокий", 
    DateTime.Now.AddDays(5)
);

delegateTask.ReassignTo("Анна Петрова");

Console.WriteLine("\nДетали задачи:");
Console.WriteLine(delegateTask.GetTaskDetails());

Console.WriteLine("\nОтметка о выполнении...");
delegateTask.MarkAsComplete();
Console.WriteLine();

Console.WriteLine("Командная задача");
Console.WriteLine("".PadRight(40, '-'));

TeamTask teamTask = new TeamTask(
    102,
    "Реализовать модуль аутентификации",
    "Критический",
    "Бэкенд команда"
);

teamTask.ReassignTo("Игорь Смирнов");

Console.WriteLine("\nДетали задачи:");
Console.WriteLine(teamTask.GetTaskDetails());

Console.WriteLine("\nПереназначение задачи...");
teamTask.ReassignTo("Мария Козлова");
Console.WriteLine();

Console.WriteLine("Исследовательская задача");
Console.WriteLine("".PadRight(40, '-'));

ResearchTask researchTask = new ResearchTask(
    103,
    "Анализ поведения пользователей",
    "Средний",
    "Google Analytics, Опросы пользователей, Лог-файлы",
    "Исследование пользовательского опыта"
);

researchTask.ReassignTo("Дмитрий Волков");

Console.WriteLine("\nДетали задачи:");
Console.WriteLine(researchTask.GetTaskDetails());

Console.WriteLine("\nОтметка о выполнении исследовательской задачи...");
researchTask.MarkAsComplete();
Console.WriteLine();

Console.WriteLine("Различные типы задач");
Console.WriteLine("".PadRight(40, '-'));

List<Task> tasks = new List<Task>
{
    delegateTask,
    teamTask,
    researchTask,
    new DelegateTask(104, "Исправить ошибку входа", "Высокий", DateTime.Now.AddDays(2))
};

Console.WriteLine("\nОбработка всех задач:\n");

foreach (Task task in tasks)
{
    Console.WriteLine($"Задача: {task.TaskName}");
    Console.WriteLine($"Тип: {task.GetType().Name}");
    
    task.ReassignTo($"Участник команды (Задача {task.TaskId})");
    
    Console.WriteLine("\nДетали:");
    Console.WriteLine(task.GetTaskDetails());
    Console.WriteLine();
    
    task.MarkAsComplete();
    Console.WriteLine(new string('-', 40));
    Console.WriteLine();
}

Console.WriteLine("Сравнение типов задач");
Console.WriteLine("".PadRight(40, '-'));

Console.WriteLine("\nДелегатская задача: Фокус на сроках и индивидуальной ответственности");
Console.WriteLine("Командная задача: Фокус на командной работе и переназначении");
Console.WriteLine("Исследовательская задача: Фокус на источниках данных и результатах исследования");

Console.WriteLine("\n" + "".PadRight(60, '='));
Console.WriteLine("Демонстрация завершена");
Console.WriteLine("".PadRight(60, '='));
```
