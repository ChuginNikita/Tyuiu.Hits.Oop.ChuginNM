```
public interface IReportable
{
    void GenerateReport();
}

public interface IPrioritizable
{
    void ChangePriority(string newPriority);
}

public interface IAuditable
{
    void LogAction(string action);
}

public class Task : IReportable, IPrioritizable, IAuditable
{
    private int taskId;
    private string taskName;
    private string priority;
    private string description;
    private string createdBy;

    public int TaskId
    {
        get { return taskId; }
        set { taskId = value; }
    }

    public string TaskName
    {
        get { return taskName; }
        set { taskName = value; }
    }

    public string Priority
    {
        get { return priority; }
        set { priority = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public string CreatedBy
    {
        get { return createdBy; }
        set { createdBy = value; }
    }

    public bool IsCompleted { get; protected set; }
    public string AssignedTo { get; protected set; }
    public DateTime CreatedDate { get; set; }
    public DateTime LastModified { get; protected set; }

    public Task()
    {
        TaskId = 0;
        TaskName = "Безымянная задача";
        Priority = "Средний";
        Description = "Описание отсутствует";
        CreatedBy = "Система";
        CreatedDate = DateTime.Now;
        LastModified = DateTime.Now;
        IsCompleted = false;
        AssignedTo = "Не назначен";
    }

    public Task(int taskId, string taskName, string priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        Description = "Описание отсутствует";
        CreatedBy = "Система";
        CreatedDate = DateTime.Now;
        LastModified = DateTime.Now;
        IsCompleted = false;
        AssignedTo = "Не назначен";
    }

    public Task(int taskId, string taskName, string priority, string description, string createdBy)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        Description = description;
        CreatedBy = createdBy;
        CreatedDate = DateTime.Now;
        LastModified = DateTime.Now;
        IsCompleted = false;
        AssignedTo = "Не назначен";
    }

    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        LastModified = DateTime.Now;
        LogAction($"Задача '{TaskName}' завершена");
    }

    public virtual string GetTaskDetails()
    {
        return $"ID: {TaskId}, Название: {TaskName}, Приоритет: {Priority}, Статус: {(IsCompleted ? "Выполнена" : "В процессе")}, Назначена: {AssignedTo}, Автор: {CreatedBy}, Создана: {CreatedDate:yyyy-MM-dd}";
    }

    public virtual void ReassignTo(string newAssignee)
    {
        AssignedTo = newAssignee;
        LastModified = DateTime.Now;
        LogAction($"Задача переназначена на {newAssignee}");
    }

    public void Interact(Task otherTask)
    {
        Console.WriteLine($"Задача '{TaskName}' взаимодействует с задачей '{otherTask.TaskName}'.");
        if (!otherTask.IsCompleted)
        {
            Console.WriteLine($"Задача '{otherTask.TaskName}' всё ещё в процессе.");
        }
        else
        {
            Console.WriteLine($"Задача '{otherTask.TaskName}' уже выполнена.");
        }
    }

    public virtual void GenerateReport()
    {
        Console.WriteLine($"Отчёт по задаче '{TaskName}':");
        Console.WriteLine($"  Создана: {CreatedDate:yyyy-MM-dd}");
        Console.WriteLine($"  Последнее изменение: {LastModified:yyyy-MM-dd}");
        Console.WriteLine($"  Статус: {(IsCompleted ? "Выполнена" : "В процессе")}");
    }

    public virtual void ChangePriority(string newPriority)
    {
        string oldPriority = Priority;
        Priority = newPriority;
        LastModified = DateTime.Now;
        Console.WriteLine($"Приоритет задачи '{TaskName}' изменён с '{oldPriority}' на '{newPriority}'.");
    }

    public virtual void LogAction(string action)
    {
        Console.WriteLine($"[{DateTime.Now:yyyy-MM-dd HH:mm}] Задача '{TaskName}': {action}");
    }

    public void UpdateDescription(string newDescription)
    {
        Description = newDescription;
        LastModified = DateTime.Now;
    }

    public void DisplayDescription()
    {
        Console.WriteLine($"Описание задачи '{TaskName}': {Description}");
    }
}

public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }
    public string DelegatedBy { get; set; }
    public bool IsUrgent { get; set; }
    public int EstimatedHours { get; set; }

    public DelegateTask() : base()
    {
        DueDate = DateTime.Now.AddDays(7);
        DelegatedBy = "Не указан";
        IsUrgent = false;
        EstimatedHours = 1;
    }

    public DelegateTask(int taskId, string taskName, string priority, DateTime dueDate)
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
        DelegatedBy = "Не указан";
        IsUrgent = false;
        EstimatedHours = 1;
    }

    public DelegateTask(int taskId, string taskName, string priority, DateTime dueDate, string delegatedBy, bool isUrgent, int estimatedHours)
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
        DelegatedBy = delegatedBy;
        IsUrgent = isUrgent;
        EstimatedHours = estimatedHours;
    }

    public override void MarkAsComplete()
    {
        IsCompleted = true;
        LastModified = DateTime.Now;
        Console.WriteLine($"Делегатская задача '{TaskName}' выполнена. Срок: {DueDate:yyyy-MM-dd}");
        LogAction($"Делегатская задача завершена. Делегирована: {DelegatedBy}");
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Срок: {DueDate:yyyy-MM-dd}, Делегирована: {DelegatedBy}, Срочная: {IsUrgent}, Оценка: {EstimatedHours} ч";
    }

    public override void GenerateReport()
    {
        base.GenerateReport();
        Console.WriteLine($"  Срок выполнения: {DueDate:yyyy-MM-dd}");
        Console.WriteLine($"  Делегирована: {DelegatedBy}");
        Console.WriteLine($"  Срочность: {(IsUrgent ? "Срочная" : "Обычная")}");
        Console.WriteLine($"  Оценка времени: {EstimatedHours} ч");
    }

    public override void ChangePriority(string newPriority)
    {
        base.ChangePriority(newPriority);
        if (newPriority == "Высокий" || newPriority == "Критический")
        {
            IsUrgent = true;
            Console.WriteLine($"  Задача '{TaskName}' помечена как срочная.");
        }
    }

    public void CheckDeadline()
    {
        if (DateTime.Now > DueDate)
        {
            Console.WriteLine($"Задача '{TaskName}' просрочена!");
        }
        else
        {
            Console.WriteLine($"Задача '{TaskName}' в срок. Дней осталось: {(DueDate - DateTime.Now).Days}");
        }
    }

    public void ExtendDeadline(int days)
    {
        DueDate = DueDate.AddDays(days);
        LastModified = DateTime.Now;
        Console.WriteLine($"Срок задачи '{TaskName}' продлён на {days} дней. Новый срок: {DueDate:yyyy-MM-dd}");
    }
}

public class TeamTask : Task
{
    public string TeamName { get; set; }
    public int TeamSize { get; set; }
    public string TeamLead { get; set; }
    public string ProjectCode { get; set; }

    public TeamTask() : base()
    {
        TeamName = "Общая команда";
        TeamSize = 1;
        TeamLead = "Не указан";
        ProjectCode = "N/A";
    }

    public TeamTask(int taskId, string taskName, string priority, string teamName)
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
        TeamSize = 1;
        TeamLead = "Не указан";
        ProjectCode = "N/A";
    }

    public TeamTask(int taskId, string taskName, string priority, string teamName, int teamSize, string teamLead, string projectCode)
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
        TeamSize = teamSize;
        TeamLead = teamLead;
        ProjectCode = projectCode;
    }

    public override void ReassignTo(string newAssignee)
    {
        string oldAssignee = AssignedTo;
        AssignedTo = newAssignee;
        LastModified = DateTime.Now;
        Console.WriteLine($"Задача '{TaskName}' переназначена с {oldAssignee} на {newAssignee} в команде '{TeamName}'.");
        LogAction($"Переназначена с {oldAssignee} на {newAssignee} в команде '{TeamName}'");
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Команда: {TeamName}, Размер команды: {TeamSize}, Лидер: {TeamLead}, Проект: {ProjectCode}";
    }

    public override void GenerateReport()
    {
        base.GenerateReport();
        Console.WriteLine($"  Команда: {TeamName}");
        Console.WriteLine($"  Размер команды: {TeamSize}");
        Console.WriteLine($"  Лидер: {TeamLead}");
        Console.WriteLine($"  Код проекта: {ProjectCode}");
    }

    public override void ChangePriority(string newPriority)
    {
        base.ChangePriority(newPriority);
        Console.WriteLine($"  Уведомление отправлено команде '{TeamName}' о смене приоритета.");
    }

    public void AssignTeamMember(string member)
    {
        TeamSize++;
        LastModified = DateTime.Now;
        Console.WriteLine($"{member} назначен в команду '{TeamName}'. Новый размер команды: {TeamSize}");
    }

    public void RemoveTeamMember(string member)
    {
        if (TeamSize > 1)
        {
            TeamSize--;
            LastModified = DateTime.Now;
            Console.WriteLine($"{member} удалён из команды '{TeamName}'. Новый размер команды: {TeamSize}");
        }
        else
        {
            Console.WriteLine($"Нельзя удалить последнего члена команды '{TeamName}'.");
        }
    }

    public void ChangeTeamLead(string newLead)
    {
        string oldLead = TeamLead;
        TeamLead = newLead;
        LastModified = DateTime.Now;
        Console.WriteLine($"Лидер команды '{TeamName}' изменён с {oldLead} на {newLead}.");
    }
}

public class ResearchTask : Task
{
    public string DataSource { get; set; }
    public string ResearchMethod { get; set; }
    public int SampleSize { get; set; }
    public bool IsPeerReviewed { get; set; }

    public ResearchTask() : base()
    {
        DataSource = "Открытые источники";
        ResearchMethod = "Наблюдение";
        SampleSize = 0;
        IsPeerReviewed = false;
    }

    public ResearchTask(int taskId, string taskName, string priority, string dataSource)
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
        ResearchMethod = "Наблюдение";
        SampleSize = 0;
        IsPeerReviewed = false;
    }

    public ResearchTask(int taskId, string taskName, string priority, string dataSource, string researchMethod, int sampleSize, bool isPeerReviewed)
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
        ResearchMethod = researchMethod;
        SampleSize = sampleSize;
        IsPeerReviewed = isPeerReviewed;
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Источник данных: {DataSource}, Метод: {ResearchMethod}, Выборка: {SampleSize}, Рецензировано: {IsPeerReviewed}";
    }

    public override void GenerateReport()
    {
        base.GenerateReport();
        Console.WriteLine($"  Источник данных: {DataSource}");
        Console.WriteLine($"  Метод исследования: {ResearchMethod}");
        Console.WriteLine($"  Размер выборки: {SampleSize}");
        Console.WriteLine($"  Рецензировано: {(IsPeerReviewed ? "Да" : "Нет")}");
    }

    public override void MarkAsComplete()
    {
        IsCompleted = true;
        LastModified = DateTime.Now;
        Console.WriteLine($"Исследовательская задача '{TaskName}' выполнена.");
        LogAction($"Исследование завершено. Источник: {DataSource}");
    }

    public void AnalyzeData(Task relatedTask)
    {
        Console.WriteLine($"Анализ данных из источника '{DataSource}' по задаче '{TaskName}'.");
        Console.WriteLine($"Связанная задача: '{relatedTask.TaskName}'.");
        LogAction($"Проведён анализ данных по задаче '{relatedTask.TaskName}'");
    }

    public void AddDataSource(string source)
    {
        DataSource += ", " + source;
        LastModified = DateTime.Now;
        Console.WriteLine($"Добавлен источник данных '{source}' к задаче '{TaskName}'.");
    }

    public void MarkAsPeerReviewed()
    {
        IsPeerReviewed = true;
        LastModified = DateTime.Now;
        Console.WriteLine($"Задача '{TaskName}' прошла рецензирование.");
    }
}

public class ScientificResearchTask : ResearchTask
{
    public string ScientificField { get; set; }
    public string Hypothesis { get; set; }
    public bool IsPublished { get; set; }

    public ScientificResearchTask() : base()
    {
        ScientificField = "Общая наука";
        Hypothesis = "Гипотеза не сформулирована";
        IsPublished = false;
    }

    public ScientificResearchTask(int taskId, string taskName, string priority, string dataSource, string researchMethod, int sampleSize, bool isPeerReviewed, string scientificField, string hypothesis)
        : base(taskId, taskName, priority, dataSource, researchMethod, sampleSize, isPeerReviewed)
    {
        ScientificField = scientificField;
        Hypothesis = hypothesis;
        IsPublished = false;
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $", Область науки: {ScientificField}, Гипотеза: {Hypothesis}, Опубликовано: {IsPublished}";
    }

    public override void GenerateReport()
    {
        base.GenerateReport();
        Console.WriteLine($"  Область науки: {ScientificField}");
        Console.WriteLine($"  Гипотеза: {Hypothesis}");
        Console.WriteLine($"  Опубликовано: {(IsPublished ? "Да" : "Нет")}");
    }

    public void Publish()
    {
        IsPublished = true;
        LastModified = DateTime.Now;
        Console.WriteLine($"Научная работа '{TaskName}' опубликована в области '{ScientificField}'.");
    }

    public void FormulateHypothesis(string hypothesis)
    {
        Hypothesis = hypothesis;
        LastModified = DateTime.Now;
        Console.WriteLine($"Гипотеза задачи '{TaskName}' сформулирована: {hypothesis}");
    }
}

        DelegateTask delegateTask = new DelegateTask(1, "Документация", "Высокий", DateTime.Now.AddDays(5), "Руководитель", false, 8);
        TeamTask teamTask = new TeamTask(2, "Аутентификация", "Критический", "Бэкенд", 5, "Игорь", "PRJ-2024");
        ResearchTask researchTask = new ResearchTask(3, "Анализ данных", "Средний", "Google Analytics", "Статистический анализ", 1000, false);
        ScientificResearchTask scienceTask = new ScientificResearchTask(4, "Квантовое исследование", "Высокий", "CERN", "Эксперимент", 5000, true, "Физика", "Гипотеза о поведении частиц");

        Console.WriteLine("НАЧАЛЬНОЕ СОСТОЯНИЕ:");
        Console.WriteLine(delegateTask.GetTaskDetails());
        Console.WriteLine(teamTask.GetTaskDetails());
        Console.WriteLine(researchTask.GetTaskDetails());
        Console.WriteLine(scienceTask.GetTaskDetails());

        Console.WriteLine();

        Console.WriteLine("НАЗНАЧЕНИЕ:");
        delegateTask.ReassignTo("Анна");
        teamTask.ReassignTo("Игорь");
        researchTask.ReassignTo("Дмитрий");
        scienceTask.ReassignTo("Елена");

        Console.WriteLine();

        Console.WriteLine("ДЕТАЛИ ПОСЛЕ НАЗНАЧЕНИЯ:");
        Console.WriteLine(delegateTask.GetTaskDetails());
        Console.WriteLine(teamTask.GetTaskDetails());
        Console.WriteLine(researchTask.GetTaskDetails());
        Console.WriteLine(scienceTask.GetTaskDetails());

        Console.WriteLine();

        Console.WriteLine("ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ:");
        delegateTask.Interact(teamTask);
        teamTask.Interact(researchTask);
        researchTask.AnalyzeData(delegateTask);
        scienceTask.AnalyzeData(teamTask);

        Console.WriteLine();

        Console.WriteLine("СПЕЦИФИЧЕСКИЕ МЕТОДЫ:");
        delegateTask.CheckDeadline();
        delegateTask.ExtendDeadline(3);
        teamTask.AssignTeamMember("Петр");
        teamTask.RemoveTeamMember("Петр");
        teamTask.ChangeTeamLead("Мария");
        researchTask.AddDataSource("Опросы");
        researchTask.MarkAsPeerReviewed();
        scienceTask.FormulateHypothesis("Новая гипотеза о поведении частиц");
        scienceTask.Publish();

        Console.WriteLine();

        Console.WriteLine("ИЗМЕНЕНИЕ ПРИОРИТЕТА:");
        delegateTask.ChangePriority("Критический");
        teamTask.ChangePriority("Высокий");
        researchTask.ChangePriority("Низкий");

        Console.WriteLine();

        Console.WriteLine("ОТЧЁТЫ:");
        delegateTask.GenerateReport();
        Console.WriteLine();
        teamTask.GenerateReport();
        Console.WriteLine();
        researchTask.GenerateReport();
        Console.WriteLine();
        scienceTask.GenerateReport();

        Console.WriteLine();

        Console.WriteLine("ВЫПОЛНЕНИЕ ЗАДАЧ:");
        delegateTask.MarkAsComplete();
        teamTask.MarkAsComplete();
        researchTask.MarkAsComplete();
        scienceTask.MarkAsComplete();

        Console.WriteLine();

        Console.WriteLine("ФИНАЛЬНОЕ СОСТОЯНИЕ:");
        Console.WriteLine(delegateTask.GetTaskDetails());
        Console.WriteLine(teamTask.GetTaskDetails());
        Console.WriteLine(researchTask.GetTaskDetails());
        Console.WriteLine(scienceTask.GetTaskDetails());
```
